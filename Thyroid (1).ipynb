```python
# Randomly assign mean values to some missing entries in the dataset

# 2604 missing entries in T3 column
Random_AVG_T3 = np.random.uniform(Thy_data['T3'].mean() - Thy_data['T3'].std(), Thy_data['T3'].mean() + Thy_data['T3'].std(), 2604)

# 809 missing entries in T4U column
Random_AVG_T4U = np.random.uniform(Thy_data['T4U'].mean() - Thy_data['T4U'].std(), Thy_data['T4U'].mean() + Thy_data['T4U'].std(), 809)

# 842 missing entries in TSH column
Random_AVG_TSH = np.random.uniform(Thy_data['TSH'].mean() - Thy_data['TSH'].std(), Thy_data['TSH'].mean() + Thy_data['TSH'].std(), 842)
```

```python
# Insert Random value in dataset column T3
Thy_data['T3'][Thy_data['T3'].isnull()]=Random_AVG_T3

# Insert Random value in dataset column T4U
Thy_data['T4U'][Thy_data['T4U'].isnull()]=Random_AVG_T4U

# Insert Random value in dataset column T4U
Thy_data['TSH'][Thy_data['TSH'].isnull()]=Random_AVG_TSH
```

```python
Thy_data = Thy_data.drop('patient_id',axis=1)
#dropping the irrelevant data columns
```

```python
X = Thy_data.drop(columns=['target'])
y = Thy_data['target']
#This code splits the data into two parts: one part contains all the information
#we want to use to make predictions (X), and the other part is the actual outcome
#we want to predict (y).
```

```python
# Defining variables to train and test
from sklearn.model_selection import train_test_split
# Splitting the data into training and testing sets without stratification
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size=0.3, random_state=45)
```

```python
from sklearn.svm import SVC
model = SVC(kernel='linear')
#importing svm to implement the prediction model
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score


# ... (Your previous code to load and preprocess data) ...

# Identify categorical features
categorical_features = X.select_dtypes(include=['object']).columns.tolist()

# Create a ColumnTransformer to apply OneHotEncoder to categorical features
preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', X.select_dtypes(exclude=['object']).columns.tolist()),
        ('cat', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), categorical_features),
    ])

# Create a pipeline with preprocessor and SVM model
model_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', SVC(kernel='linear')),
])

# Split the data into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size=0.3, random_state=45)

# Fit the pipeline to the training data
model_pipeline.fit(X_train, Y_train)

# Make predictions on the training data
X_train_prediction = model_pipeline.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print('Accuracy score of training data:: ', training_data_accuracy)
```

```python
from sklearn.metrics import classification_report

# Make predictions on the test data
X_test_prediction = model_pipeline.predict(X_test)

# Print the classification report for the training data
print("Classification Report for Training Data:")
print(classification_report(Y_train, X_train_prediction))

# Print the classification report for the test data
print("Classification Report for Test Data:")
print(classification_report(Y_test, X_test_prediction))
```

```python
# Calculate the accuracy score for the test data
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print('Accuracy score of test data: ', test_data_accuracy)
```
